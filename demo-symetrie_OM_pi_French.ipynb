# Symétrie moléculaire et orbitales π

<div class="intro">

<span style="color:blue"><u>**Pourquoi la symétrie ?**</u></span>

Beaucoup de molécules sont symétriques. Si on fait tourner la molécule d'eau de 180° autour de l'axe qui passe par l'oxygène et coupe l'angle H–O–H en deux, les deux hydrogènes échangent leurs places : la molécule obtenue est indiscernable de la molécule de départ. On dit que cette rotation est une **opération de symétrie** de la molécule.

La symétrie n'est pas qu'une curiosité géométrique. Elle permet de prévoir, souvent sans aucun calcul :
- quelles orbitales atomiques peuvent se combiner pour former des orbitales moléculaires (OM), et lesquelles ne le peuvent pas ;
- quelles OM sont **dégénérées**, c'est-à-dire de même énergie ;
- combien de bandes on observera dans un spectre infrarouge ou Raman ;
- si une molécule est chirale.

<span style="color:blue"><u>**Les opérations de symétrie**</u></span>

Il existe cinq types d'opérations de symétrie, chacune associée à un **élément de symétrie** :

| Opération | Notation | Élément de symétrie |
|---|---|---|
| identité | E | — |
| rotation de 360°/n | C<sub>n</sub> | axe de symétrie |
| réflexion | σ | plan de symétrie |
| inversion | i | centre d'inversion |
| rotation-réflexion | S<sub>n</sub> | axe impropre |

L'ensemble des opérations de symétrie d'une molécule forme son **groupe ponctuel**, noté par exemple C<sub>2v</sub> pour l'eau ou D<sub>6h</sub> pour le benzène. Chaque groupe a sa **table de caractères**, qui résume la façon dont les objets attachés à la molécule (atomes, orbitales, vibrations...) se transforment par les opérations du groupe.

<span style="color:blue"><u>**Ce que fait Python**</u></span>

Une opération de symétrie transforme les coordonnées des atomes : c'est une matrice 3×3. Python applique cette matrice et repère où va chaque atome : c'est le **suivi des atomes**. Tout le reste en découle. Le module `symmetry` de `pyphyschemtools` fait cette comptabilité pour vous, ce qui permet de se concentrer sur la chimie.

### Corrigé

````python
#################################################################################
t4pPC.centerTitle("1. Orientation de l'éthylène")
ethylene = MolSym.from_molView(molView(get_ppct_data("Symmetry/ethylene_model.xyz"), source="file", display_now=False), "D2h", lang="Fr")
ethylene.orientation()
print(np.round(ethylene.coords, 3))
display(Markdown(r'''
L'éthylène est dans le plan **yz**, et la liaison C=C est selon l'axe **z** : les deux carbones ont x = y = 0. L'axe x est perpendiculaire au plan de la molécule.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("2. Table de caractères de D2h")
display(ethylene.character_table())
display(Markdown(r'''
Le groupe D<sub>2h</sub> contient **8 opérations** : E, trois rotations C<sub>2</sub> autour des axes z, y et x, l'inversion i, et trois réflexions par rapport aux plans xy, xz et yz.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("3. Éléments de symétrie de l'éthylène")
ethylene.show_elements(labels=True)
````

````python
ethylene.show_elements("C2(z)")
ethylene.show_elements(["C2(y)", "C2(x)"])
ethylene.show_elements("i")
ethylene.show_elements(["σ(xy)", "σ(xz)", "σ(yz)"])
````

````python
display(Markdown(r'''
- **C<sub>2</sub>(z)** : axe qui porte la liaison C=C ; il passe par les deux carbones.
- **C<sub>2</sub>(y)** : axe dans le plan de la molécule, perpendiculaire à C=C, passant par le milieu de la liaison.
- **C<sub>2</sub>(x)** : axe perpendiculaire au plan de la molécule, passant par le milieu de la liaison C=C.
- **i** : centre d'inversion au milieu de la liaison C=C.
- **σ(yz)** : plan de la molécule. **σ(xz)** : plan perpendiculaire à la molécule, contenant la liaison C=C. **σ(xy)** : plan perpendiculaire à la liaison C=C, passant par son milieu.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("4. Éléments de symétrie du benzène")
benzene = MolSym.from_molView(molView(get_ppct_data("Symmetry/benzene_model.xyz"), source="file", display_now=False), "D6h", lang="Fr")
benzene.orientation()
display(benzene.character_table())
benzene.show_elements()
display(Markdown(r'''
Le benzène est dans le plan **xy**, et l'axe C<sub>6</sub> est selon **z**, perpendiculaire au cycle. Le groupe D<sub>6h</sub> contient **24 opérations** : il suffit d'additionner les nombres placés devant les noms des classes (1 + 2 + 2 + 1 + 3 + 3 + 1 + 2 + 2 + 1 + 3 + 3 = 24).
'''))
````

````python
#################################################################################
t4pPC.centerTitle("5. C2' et C2'', σv et σd")
benzene.show_elements(["C2'", "C2''"], labels=True)
benzene.show_elements(["σv", "σd"])
display(Markdown(r'''
Les trois axes **C<sub>2</sub>'** passent par deux atomes de carbone opposés (et par les H qu'ils portent). Les trois axes **C<sub>2</sub>''** passent par les milieux de deux liaisons C–C opposées. Une rotation C<sub>6</sub> ne peut pas transformer un axe qui passe par des atomes en un axe qui passe par des milieux de liaisons : ce sont donc deux classes différentes.

De même, les plans **σ<sub>v</sub>** contiennent l'axe C<sub>6</sub> et passent par des atomes, les plans **σ<sub>d</sub>** contiennent l'axe C<sub>6</sub> et passent par des milieux de liaisons.
'''))
````

## 2. Comment les atomes s'échangent

<div class="ex">

Une opération de symétrie envoie chaque atome sur un atome du même élément. La méthode `track()` indique, pour chaque atome, l'atome sur lequel il arrive. Par exemple, `ethylene.track("C2(z)")` affiche `C0 → C0, C1 → C1, H2 → H3...`.

1. Pour l'éthylène, utilisez `track()` pour chacune des opérations C<sub>2</sub>(z), C<sub>2</sub>(y), C<sub>2</sub>(x), i, σ(xy), σ(xz) et σ(yz).
2. Visualisez l'effet de C<sub>2</sub>(y) et de i avec `show_operation()` : la molécule de départ est à gauche, la molécule après l'opération à droite, chaque atome gardant son numéro.
3. Pour chaque opération, comptez les atomes qui **restent en place**. Reportez ces nombres dans un tableau, sous les colonnes de la table de caractères. Comparez avec le résultat de `gamma_atoms()`.
4. Un atome de carbone peut-il être envoyé sur un atome d'hydrogène ? Quels atomes sont **équivalents** par symétrie ? Vérifiez avec `equivalent_atoms()`.
5. Pour le benzène, utilisez `track()` pour C<sub>6</sub>, C<sub>2</sub>' et σh. Que fait la réflexion σh ? Pourquoi aucun atome ne bouge-t-il ?

</div>
<br>
<div class="hint">

Les noms des opérations à donner à `track()` et à `show_operation()` sont ceux des colonnes de la table de caractères, sans le nombre placé devant : `"C2(y)"`, `"σ(xz)"`, `"C2'"`... Pour une classe qui contient plusieurs opérations, comme 3C<sub>2</sub>', le module utilise l'une d'entre elles.

</div>

### Corrigé

````python
#################################################################################
t4pPC.centerTitle("1. Suivi des atomes de l'éthylène")
ethylene.track("C2(z)")
ethylene.track("C2(y)")
ethylene.track("C2(x)")
ethylene.track("i")
ethylene.track("σ(xy)")
ethylene.track("σ(xz)")
ethylene.track("σ(yz)")
````

````python
#################################################################################
t4pPC.centerTitle("2. Visualisation")
ethylene.show_operation("C2(y)")
ethylene.show_operation("i")
````

````python
#################################################################################
t4pPC.centerTitle("3. Atomes qui restent en place")
ethylene.gamma_atoms()
display(Markdown(r'''
- **E** et **σ(yz)** (plan de la molécule) laissent les 6 atomes en place.
- **C<sub>2</sub>(z)** et **σ(xz)** laissent en place les deux carbones, situés sur l'axe z et dans le plan xz, mais échangent les hydrogènes.
- **C<sub>2</sub>(y)**, **C<sub>2</sub>(x)**, **i** et **σ(xy)** déplacent tous les atomes : par exemple, i échange les deux carbones.

La ligne obtenue, (6, 2, 0, 0, 0, 0, 2, 6), est appelée la représentation **Γ<sub>atomes</sub>**.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("4. Atomes équivalents")
print(ethylene.equivalent_atoms())
display(Markdown(r'''
Une opération de symétrie laisse la molécule **inchangée** : un atome de carbone est toujours envoyé sur un atome de carbone, jamais sur un hydrogène. Il y a deux ensembles d'atomes équivalents : les deux carbones (C0, C1) et les quatre hydrogènes (H2 à H5). Deux atomes équivalents ont exactement le même environnement chimique : par exemple, les quatre H de l'éthylène donnent un seul signal en RMN.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("5. Benzène")
benzene.track("C6")
benzene.track("C2'")
benzene.track("σh")
display(Markdown(r'''
C<sub>6</sub> fait « tourner » les atomes d'un cran autour du cycle. C<sub>2</sub>' laisse en place les deux C et les deux H situés sur l'axe, et échange les autres deux à deux. La réflexion **σh** se fait par rapport au plan de la molécule : tous les atomes sont dans ce plan, aucun ne bouge.
'''))
````

h_eth = HMO("C=C")

Affichez le diagramme d'énergie (`plot_energy_diagram()`) et les OM (`plot_all_mo()`). Les énergies s'écrivent E = α + xβ, et les valeurs de x sont dans `h_eth.E_MOs`.

4. La méthode `orbital_symmetries()` donne la symétrie de chaque OM calculée :

```python
ethylene.orbital_symmetries(h_eth.c_MOs, h_eth.E_MOs, atoms=[0, 1], shell="px")
```

Quelle est la symétrie de l'OM liante π ? Et de l'OM antiliante π\* ?
5. Les indices g et u indiquent le comportement de l'OM par l'inversion i : g (*gerade*, pair) si l'OM est inchangée, u (*ungerade*, impair) si elle change de signe. En regardant le dessin des OM, justifiez que π est u et π\* est g.

</div>
<br>
<div class="hint">

`atoms=[0, 1]` indique que les coefficients de Hückel portent sur les atomes C0 et C1 de la molécule, dans cet ordre. Pour une orbitale p, l'inversion transforme le lobe + en lobe −.

</div>

### Corrigé

````python
#################################################################################
t4pPC.centerTitle("1. Orbitales p du système π")
display(Markdown(r'''
L'éthylène est dans le plan yz : l'orbitale perpendiculaire au plan est **p<sub>x</sub>**.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("2. Symétrie des combinaisons de p_x")
ethylene.gamma_orbitals({"C": "px"})
display(Markdown(r'''
On obtient **deux** représentations irréductibles, B<sub>2g</sub> et B<sub>3u</sub>, de dimension 1 : deux orbitales p<sub>x</sub> donnent **deux** OM π.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("3. Calcul de Hückel")
h_eth = HMO("C=C")
print(h_eth.E_MOs)
h_eth.plot_energy_diagram()
h_eth.plot_all_mo()
````

````python
#################################################################################
t4pPC.centerTitle("4. Symétrie des OM")
ethylene.orbital_symmetries(h_eth.c_MOs, h_eth.E_MOs, atoms=[0, 1], shell="px")
display(Markdown(r'''
- OM liante **π** (x = 1, E = α + β) : symétrie **B<sub>3u</sub>** ;
- OM antiliante **π\*** (x = −1, E = α − β) : symétrie **B<sub>2g</sub>**.

On retrouve les deux représentations prévues à la question 2, sans avoir fait le calcul de Hückel.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("5. Pourquoi u et g ?")
display(Markdown(r'''
L'inversion envoie C0 sur C1 et transforme chaque lobe + en lobe −.
- Dans **π**, les deux orbitales p<sub>x</sub> sont en phase : après l'inversion, on retrouve la même OM avec un signe −. Elle est **u**.
- Dans **π\***, les deux orbitales sont en opposition de phase : les deux changements (échange des atomes et changement de signe des lobes) se compensent. L'OM est inchangée : elle est **g**.
'''))
````

## 4. Symétrie des orbitales π du benzène

<div class="ex">

1. Quelle orbitale p des carbones du benzène participe au système π ?
2. Utilisez `gamma_orbitals()` avec cette orbitale. Quelles sont les représentations irréductibles obtenues ? La première lettre indique la dimension de la représentation : 1 pour A et B, 2 pour E. Vérifiez qu'on obtient bien six OM π au total.
3. Faites le calcul de Hückel du benzène avec `HMO("c1ccccc1")`, affichez le diagramme d'énergie et les OM, puis donnez la symétrie de chaque OM avec `orbital_symmetries()`. Les six carbones du fichier `benzene_model.xyz` sont les atomes 0 à 5, numérotés dans le même ordre que dans le calcul de Hückel.
4. Certaines OM ont la même énergie : elles sont **dégénérées**. Quel est le lien avec leur symétrie ?
5. Essayez de trouver la symétrie d'une seule des deux OM de x = 1 :

```python
benzene.orbital_symmetry(h_benz.c_MOs[:, 1], atoms=[0, 1, 2, 3, 4, 5])
```

Que se passe-t-il ? Pourquoi ?
6. Le benzène a 6 électrons π. Quelle est la symétrie de la HO et de la BV ?
7. Comptez le nombre de plans nodaux perpendiculaires au cycle de chaque OM. Comment évolue-t-il avec l'énergie ?

</div>
<br>
<div class="hint">

Les deux OM dégénérées de x = 1 sont les colonnes 1 et 2 de `h_benz.c_MOs` : on les donne ensemble avec `h_benz.c_MOs[:, [1, 2]]`.

</div>

### Corrigé

````python
#################################################################################
t4pPC.centerTitle("1. et 2. Symétrie des combinaisons de p_z")
benzene.gamma_orbitals({"C": "pz"})
display(Markdown(r'''
Le benzène est dans le plan xy : le système π est construit avec les orbitales **p<sub>z</sub>**. On obtient A<sub>2u</sub> + B<sub>2g</sub> + E<sub>1g</sub> + E<sub>2u</sub>, soit 1 + 1 + 2 + 2 = **6 OM π**, autant que d'orbitales p<sub>z</sub>.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("3. Calcul de Hückel et symétrie des OM")
h_benz = HMO("c1ccccc1")
print(h_benz.E_MOs)
h_benz.plot_energy_diagram()
h_benz.plot_all_mo()
benzene.orbital_symmetries(h_benz.c_MOs, h_benz.E_MOs, atoms=[0, 1, 2, 3, 4, 5])
````

````python
#################################################################################
t4pPC.centerTitle("4. Dégénérescence")
display(Markdown(r'''
| x | E | OM | Symétrie |
|---|---|---|---|
| 2 | α + 2β | 1 | A<sub>2u</sub> |
| 1 | α + β | 2 (dégénérées) | E<sub>1g</sub> |
| −1 | α − β | 2 (dégénérées) | E<sub>2u</sub> |
| −2 | α − 2β | 1 | B<sub>2g</sub> |

Les OM dégénérées appartiennent aux représentations **E**, de dimension 2. Ce n'est pas un hasard : une rotation C<sub>6</sub> transforme l'une des deux OM en une combinaison des deux. Elles doivent donc avoir la même énergie. La dégénérescence est une conséquence de la symétrie.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("5. Une seule OM dégénérée")
try:
    benzene.orbital_symmetry(h_benz.c_MOs[:, 1], atoms=[0, 1, 2, 3, 4, 5])
except ValueError as erreur:
    print(erreur)
benzene.orbital_symmetry(h_benz.c_MOs[:, [1, 2]], atoms=[0, 1, 2, 3, 4, 5])
display(Markdown(r'''
Une seule des deux OM de x = 1 ne suffit pas : la rotation C<sub>6</sub> la transforme en une combinaison des **deux** OM dégénérées. Il faut donner les deux OM ensemble pour obtenir la représentation E<sub>1g</sub>.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("6. HO et BV")
display(Markdown(r'''
Les 6 électrons π occupent l'OM A<sub>2u</sub> (2 électrons) et les deux OM E<sub>1g</sub> (4 électrons). La **HO** est donc de symétrie **E<sub>1g</sub>** (doublement dégénérée), et la **BV** de symétrie **E<sub>2u</sub>**.
'''))
````

````python
#################################################################################
t4pPC.centerTitle("7. Plans nodaux")
display(Markdown(r'''
| Symétrie | x | Plans nodaux perpendiculaires au cycle |
|---|---|---|
| A<sub>2u</sub> | 2 | 0 |
| E<sub>1g</sub> | 1 | 1 |
| E<sub>2u</sub> | −1 | 2 |
| B<sub>2g</sub> | −2 | 3 |

Le nombre de plans nodaux augmente avec l'énergie, comme pour les OM de l'éthylène (0 pour π, 1 pour π\*). On retrouve le comportement d'une particule sur un cercle.
'''))
````
